# 02. Text Preprocessing & Bag-of-Words Feature Extraction

This notebook covers:
1. Text cleaning and token normalization.
2. Stratified train-test split (80:20).
3. Bag-of-Words feature extraction using `CountVectorizer` (max_features=3000).


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.data_loader import load_data
from src.preprocessing import preprocess_text, clean_text
from src.feature_extraction import split_data, extract_bow_features
```

### 1. Preprocessing Showcase
Let us test different cleaning modes on sample SMS messages.


```python
sample = 'URGENT! You won £1,000 cash. Call 0800-123456 or click http://prize.com now!'
print('Original: ', sample)
print('Mode A (Minimal):  ', preprocess_text(sample, mode='minimal'))
print('Mode B (Standard): ', preprocess_text(sample, mode='standard'))
print('Mode C (Stopwords):', preprocess_text(sample, mode='with_stopwords'))
```

### 2. Train-Test Split (Stratified 80:20)
Ensures equal class proportions in train and test splits.


```python
df = load_data()
X_train, X_test, y_train, y_test = split_data(df, test_size=0.2, random_state=42)
print(f'Train size: {len(X_train)} (Ham: {(y_train==0).sum()}, Spam: {(y_train==1).sum()})')
print(f'Test size:  {len(X_test)} (Ham: {(y_test==0).sum()}, Spam: {(y_test==1).sum()})')
```

### 3. Bag-of-Words Extraction (CountVectorizer)
**Crucial rule**: Fit ONLY on training data, then transform test data.


```python
X_train_vec, X_test_vec, vec = extract_bow_features(X_train, X_test, max_features=3000)
print(f'Vocabulary size: {len(vec.vocabulary_)}')
print(f'X_train shape: {X_train_vec.shape}')
print(f'X_test shape: {X_test_vec.shape}')
print('Sample features:', list(vec.vocabulary_.keys())[:20])
```
